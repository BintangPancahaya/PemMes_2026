```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score

# 1. PERSIAPAN DATASET
cancer = load_breast_cancer()
df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df['diagnosis'] = cancer.target

df['diagnosis'] = df['diagnosis'].map({0: 'M', 1: 'B'})

df.insert(0, 'id', range(1, len(df) + 1))
```

```python
# JAWABAN NO 1: Memisahkan variabel

# Kolom 'id' tidak berguna untuk analisis, jadi kita hapus
X = df.drop(columns=['id', 'diagnosis'])
y = df['diagnosis']
```

```python
# JAWABAN NO 2: Encoding pada kolom "diagnosis"

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
```

```python
# JAWABAN NO 3, 4, 5, 6

# Kita coba memilih 10 fitur terbaik sebagai awal (k=10)
pipeline = Pipeline([
    ('scaler', StandardScaler()),                                   # No 3: Standardisasi
    ('feature_selection', SelectKBest(score_func=f_classif, k=10)), # No 4: Seleksi Fitur
    ('model', LogisticRegression())                                 # No 5: Logistic Regression
    ])

pipeline.fit(X_train, y_train)

y_pred = pipeline.predict(X_test)
print(f"Akurasi Model (dengan 10 Fitur Terbaik): {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
```

```python
# JAWABAN NO 7: Analisis Fitur Terbaik

# Mengambil nama-nama fitur yang terpilih oleh SelectKBest
selector = pipeline.named_steps['feature_selection']
fitur_terpilih_mask = selector.get_support()
fitur_terpilih = X.columns[fitur_terpilih_mask]

print(f"Jumlah fitur terbaik yang digunakan: {len(fitur_terpilih)}")
print("Daftar fitur tersebut:")
for i, name in enumerate(fitur_terpilih, 1):
    print(f"{i}. {name}")
```
